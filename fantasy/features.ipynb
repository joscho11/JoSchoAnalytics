# Feature Engineering Pipeline

Inputs `raw_dataset.csv`, outputs `features_dataset.csv`. Run cells top-to-bottom.

## Setup

```python
import pandas as pd
import numpy as np

# Paths are relative to the fantasy/ directory (standard run location)
df = pd.read_csv("raw_dataset.csv")

# Make sure it's sorted correctly for rolling window calculations
df = df.sort_values(["player_id", "season", "week"]).reset_index(drop=True)

print(df.shape)
df.head()
```

## Step 1 — Target Variable

```python
# Half PPR = standard + (0.5 * receptions)
df["fantasy_points_half_ppr"] = df["fantasy_points"] + (0.5 * df["receptions"])

# Quick sanity check on a known player
jj = df[df["player_display_name"] == "Justin Jefferson"][["season", "week", "receptions", "fantasy_points", "fantasy_points_half_ppr"]].head(10)
print(jj)
```

```python
# For each player, shift their fantasy points back by 1 row within each season.
# groupby(["player_id","season"]) ensures week 18 of 2021 does NOT target week 1 of 2022.
df["target_half_ppr"] = (df.groupby(["player_id", "season"])["fantasy_points_half_ppr"]
                           .shift(-1))

# These rows (last week of each season per player) will be dropped before training
print(f"Rows with valid target: {df['target_half_ppr'].notna().sum()}")
print(f"Rows without target (last week of season): {df['target_half_ppr'].isna().sum()}")
```

## Step 2 — Rolling Features & Trends

```python
# Columns we want rolling averages for
roll_cols = [
    "fantasy_points_half_ppr",
    "targets", "receptions", "receiving_yards", "receiving_tds",
    "carries", "rushing_yards", "rushing_tds",
    "target_share", "air_yards_share", "wopr",
    "receiving_air_yards", "receiving_yards_after_catch",
    "ffo_expected_pts", "ffo_pts_diff",
]

# Rolling 3 week and 5 week averages
# shift(1) is critical — ensures we only use PAST data, not the current week
for col in roll_cols:
    df[f"{col}_roll3"] = (
        df.groupby(["player_id", "season"])[col]
        .transform(lambda x: x.shift(1).rolling(3, min_periods=1).mean())
    )
    df[f"{col}_roll5"] = (
        df.groupby(["player_id", "season"])[col]
        .transform(lambda x: x.shift(1).rolling(5, min_periods=1).mean())
    )

print(f"New shape: {df.shape}")
print("Sample rolling features:")
df[df["player_display_name"] == "Justin Jefferson"][
    ["season", "week", "fantasy_points_half_ppr",
     "fantasy_points_half_ppr_roll3", "fantasy_points_half_ppr_roll5",
     "targets_roll3", "target_share_roll3"]
].head(15)
```

```python
# Trend = 3 week average minus 5 week average
# Positive = usage trending up, Negative = trending down
trend_cols = [
    "fantasy_points_half_ppr", "targets", "target_share",
    "air_yards_share", "wopr", "carries"
]

for col in trend_cols:
    df[f"{col}_trend"] = df[f"{col}_roll3"] - df[f"{col}_roll5"]

print("Sample trend features:")
df[df["player_display_name"] == "Justin Jefferson"][
    ["season", "week", "targets_roll3", "targets_roll5", "targets_trend"]
].head(15)
```

## Step 3 — Points Allowed vs Position

```python
# First build a weekly pts allowed per team per position table
pts_allowed = (
    df.groupby(["season", "week", "opponent_team", "position"])["fantasy_points_half_ppr"]
    .sum()
    .reset_index()
    .rename(columns={
        "opponent_team": "team",
        "fantasy_points_half_ppr": "pts_scored_vs_team"
    })
)

# Rolling 4 week average pts allowed by each defense vs each position
pts_allowed = pts_allowed.sort_values(["team", "position", "season", "week"])
pts_allowed["def_pts_allowed_roll4"] = (
    pts_allowed.groupby(["team", "position", "season"])["pts_scored_vs_team"]
    .transform(lambda x: x.shift(1).rolling(4, min_periods=1).mean())
)

# Join back onto main df using opponent_team
df = df.merge(
    pts_allowed[["season", "week", "team", "position", "def_pts_allowed_roll4"]],
    left_on=["season", "week", "opponent_team", "position"],
    right_on=["season", "week", "team", "position"],
    how="left"
).drop(columns=["team_y"]).rename(columns={"team_x": "team"})

print("Sample defense features:")
df[df["player_display_name"] == "Justin Jefferson"][
    ["season", "week", "opponent_team", "def_pts_allowed_roll4"]
].head(10)
```

## Step 4 — Coach Features

```python
# Impute coach_win_pct nulls (~9.9% — coaches with < 10 career games)
# Flag new coaches first, then fill with league median
coach_median     = df[df['season'] <= 2024]['coach_win_pct'].median()
opp_coach_median = df[df['season'] <= 2024]['opp_coach_win_pct'].median()

df['is_new_coach']     = df['coach_win_pct'].isna().astype(int)
df['opp_is_new_coach'] = df['opp_coach_win_pct'].isna().astype(int)
df['coach_win_pct']     = df['coach_win_pct'].fillna(coach_median)
df['opp_coach_win_pct'] = df['opp_coach_win_pct'].fillna(opp_coach_median)

print(f'coach median: {coach_median:.4f} | opp median: {opp_coach_median:.4f}')
print(f'is_new_coach rows:     {df["is_new_coach"].sum()}')
print(f'opp_is_new_coach rows: {df["opp_is_new_coach"].sum()}')
print(f'coach_win_pct nulls remaining: {df["coach_win_pct"].isna().sum()}')
```

## Step 5 — Strength of Schedule & Team Rankings

```python
# Polars >= 1.x is strict about UTF-8 in parquet files; nflverse data
# occasionally contains non-UTF-8 strings. Fall back to pyarrow on failure.
import polars as _pl
_pl_read_parquet_orig = _pl.read_parquet
def _pl_read_parquet_lenient(source, *args, **kwargs):
    try:
        return _pl_read_parquet_orig(source, *args, **kwargs)
    except Exception:
        kwargs.setdefault('use_pyarrow', True)
        return _pl_read_parquet_orig(source, *args, **kwargs)
_pl.read_parquet = _pl_read_parquet_lenient

# ── Strength of Schedule ──────────────────────────────────────────────────────
# opp_season_win_pct : current opponent's win% going into this game (display + model)
# opp_win_pct_roll4  : rolling 4-game avg win% of opponents this team has faced (SOS context for EPA)

import nflreadpy as nfl

SEASONS = sorted(df['season'].unique().tolist())
sched_raw = nfl.load_schedules(SEASONS).to_pandas()
sched = sched_raw[sched_raw['game_type'] == 'REG'][
    ['season', 'week', 'home_team', 'away_team', 'home_score', 'away_score']
].dropna(subset=['home_score', 'away_score']).copy()

# Normalize relocated franchise abbreviations to current codes (matches df['team'])
TEAM_MAP = {
    'OAK': 'LV', 'SD': 'LAC', 'STL': 'LA',
    'JAC': 'JAX', 'ARZ': 'ARI', 'BLT': 'BAL', 'CLV': 'CLE', 'HST': 'HOU',
}
sched['home_team'] = sched['home_team'].replace(TEAM_MAP)
sched['away_team'] = sched['away_team'].replace(TEAM_MAP)

# Build one row per team per game
home = sched[['season', 'week', 'home_team', 'home_score', 'away_score']].copy()
home['team'] = home['home_team']
home['win']  = (home['home_score'] > home['away_score']).astype(int)

away = sched[['season', 'week', 'away_team', 'home_score', 'away_score']].copy()
away['team'] = away['away_team']
away['win']  = (away['away_score'] > away['home_score']).astype(int)

records = pd.concat([
    home[['season', 'week', 'team', 'win']],
    away[['season', 'week', 'team', 'win']]
]).sort_values(['team', 'season', 'week']).reset_index(drop=True)

# Win% GOING INTO each week (cumulative through prior weeks, no leakage)
records['cum_wins']   = records.groupby(['team', 'season'])['win'].cumsum()
records['cum_games']  = records.groupby(['team', 'season']).cumcount() + 1
records['win_pct']    = (records['cum_wins'] - records['win']) / (records['cum_games'] - 1)
records['win_pct']    = records['win_pct'].fillna(0.5)  # week 1: no prior record

# Rename for join onto opponent_team
opp_wp = records[['season', 'week', 'team', 'win_pct']].rename(
    columns={'team': 'opponent_team', 'win_pct': 'opp_season_win_pct'}
)

df = df.merge(opp_wp, on=['season', 'week', 'opponent_team'], how='left')
df['opp_season_win_pct'] = df['opp_season_win_pct'].fillna(0.5)

# Rolling 4-game avg of opponent win% (team-level SOS rolling)
team_sos = (
    df.drop_duplicates(subset=['team', 'season', 'week'])
    [['team', 'season', 'week', 'opp_season_win_pct']]
    .sort_values(['team', 'season', 'week'])
    .copy()
)
team_sos['opp_win_pct_roll4'] = (
    team_sos.groupby(['team', 'season'])['opp_season_win_pct']
    .transform(lambda x: x.shift(1).rolling(4, min_periods=1).mean())
)
team_sos['opp_win_pct_roll4'] = team_sos['opp_win_pct_roll4'].fillna(0.5)

df = df.merge(
    team_sos[['team', 'season', 'week', 'opp_win_pct_roll4']],
    on=['team', 'season', 'week'], how='left'
)

print(f'opp_season_win_pct nulls: {df["opp_season_win_pct"].isna().sum()}')
print(f'opp_win_pct_roll4  nulls: {df["opp_win_pct_roll4"].isna().sum()}')
print(f'Sample opp_season_win_pct range: {df["opp_season_win_pct"].min():.3f} – {df["opp_season_win_pct"].max():.3f}')
```

```python
# ── Vegas spread and game total ────────────────────────────────────────────
# total_line  : over/under (game pace signal -- higher = more plays = more opportunity)
# team_spread : from this team's perspective, negative = favored (game script signal)
# sched_raw is already loaded above from nfl.load_schedules()

spread_home = sched_raw[sched_raw['game_type'] == 'REG'][
    ['season', 'week', 'home_team', 'spread_line', 'total_line']
].copy()
spread_home.rename(columns={'home_team': 'team'}, inplace=True)
spread_home['team_spread'] = spread_home['spread_line']   # negative = home favored

spread_away = sched_raw[sched_raw['game_type'] == 'REG'][
    ['season', 'week', 'away_team', 'spread_line', 'total_line']
].copy()
spread_away.rename(columns={'away_team': 'team'}, inplace=True)
spread_away['team_spread'] = -spread_away['spread_line']  # flip to away-team perspective

spread_ctx = pd.concat([spread_home, spread_away], ignore_index=True)[
    ['season', 'week', 'team', 'total_line', 'team_spread']
]

df = df.drop(columns=['total_line', 'team_spread'], errors='ignore')
df = df.merge(spread_ctx, on=['season', 'week', 'team'], how='left')

_train_total_median = df.loc[df['season'] <= 2024, 'total_line'].median()
df['total_line']  = df['total_line'].fillna(_train_total_median)
df['team_spread'] = df['team_spread'].fillna(0.0)

print(f'total_line  null after fill: {df["total_line"].isna().sum()}')
print(f'team_spread null after fill: {df["team_spread"].isna().sum()}')
print(f'total_line  mean: {df["total_line"].mean():.1f}  range: [{df["total_line"].min():.1f}, {df["total_line"].max():.1f}]')
print(f'team_spread mean: {df["team_spread"].mean():.2f}  range: [{df["team_spread"].min():.1f}, {df["team_spread"].max():.1f}]')
```

```python
# ── Team rankings per week ────────────────────────────────────────────────────
# Rank all teams within each season/week. Rank 1 = best.
# off_epa_rank: rank 1 = highest EPA (most efficient offense)
# sos_rank:     rank 1 = highest opponent win% (toughest recent schedule)

team_week = (
    df.groupby(['team', 'season', 'week'])[['off_epa_roll4', 'opp_win_pct_roll4']]
    .mean()
    .reset_index()
)

team_week['off_epa_rank'] = (
    team_week.groupby(['season', 'week'])['off_epa_roll4']
    .rank(ascending=False, method='min', na_option='bottom')
    .astype('Int64')
)
team_week['sos_rank'] = (
    team_week.groupby(['season', 'week'])['opp_win_pct_roll4']
    .rank(ascending=False, method='min', na_option='bottom')
    .astype('Int64')
)

df = df.merge(
    team_week[['team', 'season', 'week', 'off_epa_rank', 'sos_rank']],
    on=['team', 'season', 'week'], how='left'
)

print(f'off_epa_rank nulls: {df["off_epa_rank"].isna().sum()}')
print(f'sos_rank nulls:     {df["sos_rank"].isna().sum()}')
print(f'off_epa_rank range: {int(df["off_epa_rank"].min())} – {int(df["off_epa_rank"].max())}')
print(f'sos_rank range:     {int(df["sos_rank"].min())} – {int(df["sos_rank"].max())}')
```

```python
# Drop rows where target is null (last week of each season)
df_model = df[df['target_half_ppr'].notna()].copy()

# Drop rows where all rolling features are null (week 1 of a player's first season)
df_model = df_model[df_model['fantasy_points_half_ppr_roll3'].notna()].copy()

# Define final feature columns for the model
feature_cols = [
    # identity (not used in training but useful for lookup)
    'player_id', 'player_display_name', 'position', 'team',
    'opponent_team', 'season', 'week',

    # rolling averages -- 3 week
    'fantasy_points_half_ppr_roll3', 'targets_roll3', 'receptions_roll3',
    'receiving_yards_roll3', 'receiving_tds_roll3', 'carries_roll3',
    'rushing_yards_roll3', 'rushing_tds_roll3', 'target_share_roll3',
    'air_yards_share_roll3', 'wopr_roll3', 'receiving_air_yards_roll3',
    'receiving_yards_after_catch_roll3', 'ffo_expected_pts_roll3',
    'ffo_pts_diff_roll3',

    # rolling averages -- 5 week
    'fantasy_points_half_ppr_roll5', 'targets_roll5', 'receptions_roll5',
    'receiving_yards_roll5', 'receiving_tds_roll5', 'carries_roll5',
    'rushing_yards_roll5', 'rushing_tds_roll5', 'target_share_roll5',
    'air_yards_share_roll5', 'wopr_roll5', 'receiving_air_yards_roll5',
    'receiving_yards_after_catch_roll5', 'ffo_expected_pts_roll5',
    'ffo_pts_diff_roll5',

    # trend features
    'fantasy_points_half_ppr_trend', 'targets_trend', 'target_share_trend',
    'air_yards_share_trend', 'wopr_trend', 'carries_trend',

    # snap share
    'snap_pct_roll3', 'snap_pct_roll5', 'snap_pct_trend',

    # matchup context
    'def_pts_allowed_roll4', 'implied_team_total', 'total_line', 'team_spread', 'is_home',
    'opp_season_win_pct', 'opp_win_pct_roll4',
    'off_epa_rank', 'sos_rank',

    # coach quality
    'coach_win_pct', 'opp_coach_win_pct',
    'is_new_coach', 'opp_is_new_coach',

    # team offensive metrics (rolling 4-week, own team)
    'off_epa_roll4', 'off_yards_per_play_roll4',
    'off_pass_rate_roll4', 'off_red_zone_rate_roll4',

    # opponent defensive metrics (rolling 4-week)
    'def_epa_allowed_roll4', 'def_yards_allowed_roll4',
    'def_pass_rate_faced_roll4', 'def_red_zone_allowed_roll4',

    # all-pro talent quality
    'team_allpro_weighted', 'team_offense_allpro', 'team_defense_allpro',
    'opp_allpro_weighted',  'opp_offense_allpro',  'opp_defense_allpro',

    # weather + game context
    'days_rest', 'is_dome', 'effective_wind', 'effective_temp', 'is_turf',

    # injury features
    'injury_status_score', 'practice_status_score',
    'starter_qb_availability', 'starter_rb_availability',
    'starter_wr_availability', 'starter_te_availability',
    'depth_chart_position',

    # opponent defensive injury features
    'opp_cb1_availability', 'opp_olb1_availability', 'opp_ilb1_availability',
    'opp_mlb1_availability', 'opp_de1_availability', 'opp_dt1_availability',
    'opp_fs1_availability', 'opp_ss1_availability',

    # OL injury features
    'starter_tackle_availability', 'starter_guard_availability',
    'starter_center_availability',

    # target
    'target_half_ppr',
]

# Warn about any expected columns missing from df
missing = [c for c in feature_cols if c not in df_model.columns]
if missing:
    print(f'WARNING: {len(missing)} feature cols missing from df: {missing}')

df_model = df_model[feature_cols].copy()

print(f'Final shape: {df_model.shape}')
print(f'Null counts:')
nulls = df_model.isnull().sum()
print(nulls[nulls > 0])
```

## Step 6 — Cleanup & Save

```python
df_model.to_csv(
    "features_dataset.csv",  # relative to fantasy/ directory
    index=False
)
print("✅ features_dataset.csv saved")
```
